# **key  pandas concepts**

# **1.series:**


*   a series is a one dimensional labeled array that can hold any data type: integers,floats ,strings ,or even python objects .Each element in the series has a unique labelcalled an index
*   it is often to track changes or patterns over time ,such as daily temperatures ,stock prices or sales revenue.



# **2.DataFrame:**


*   A dataframe is a two dimensional labeled data structure in pandas ,similar to a table in a database ,an excel,spreadsheet,or a SQL table.
*  It consists of rows and columns ,where:


1.   rows have indices(labels).
2.   columns have names(labels).

```python
# importing pandas
import pandas as pd

# creating dataframe
df = pd.DataFrame({'Product': ['Carrots', 'Broccoli', 'Banana', 'Banana',
                               'Beans', 'Orange', 'Broccoli', 'Banana'],
                   'Category': ['Vegetable', 'Vegetable', 'Fruit', 'Fruit',
                                'Vegetable', 'Fruit', 'Vegetable', 'Fruit'],
                   'Quantity': [8, 5, 3, 4, 5, 9, 11, 8],
                   'Amount': [270, 239, 617, 384, 626, 610, 62, 90]})
df
```

```python
#Create a Pivot Table in Pandas
pivot = df.pivot_table(index=['Product'],
                       values=['Amount'],
                       aggfunc='sum')
print(pivot)
```

```python
# creating pivot table of total
# sales category-wise aggfunc = 'sum'
pivot = df.pivot_table(index=['Category'],
                       values=['Amount'],
                       aggfunc='sum')
print(pivot)
```

```python
# 'mean', 'min'} will get median, mean and
# minimum of sales respectively
pivot = df.pivot_table(index=['Category'], values=['Amount'],
                       aggfunc={'median', 'mean', 'min'})
print(pivot)
```

# **mini project**

```python
import pandas as pd

data = {
    "Name": ["Amit", "Riya", "Karan", "Neha", "Sonal"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 92, 78, 88, 95],
    "Course": ["CS", "IT", "CS", "IT", "CS"]
}

df = pd.DataFrame(data)
print(df)
```

```python
avg_marks = df["Marks"].mean()
print("Average Marks:", avg_marks)

```

```python
course_avg = df.groupby("Course")["Marks"].mean()
print(course_avg)
```

```python
max_marks = df["Marks"].max()
print("Maximum Marks:", max_marks)
```

```python
topper = df[df["Marks"] == df["Marks"].max()]
print(topper)
```

```python
high_scorers = df[df["Marks"] > 85]
print(high_scorers)
```

```python
cs_students = df[df["Course"] == "CS"]
print(cs_students)
```

```python
above_20 = df[df["Age"] > 20]
print(above_20)
```

```python
result = df[(df["Marks"] > 85) & (df["Course"] == "CS")]
print(result)
```
