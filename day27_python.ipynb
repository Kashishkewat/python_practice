# **Lasso Regression**

**Lasso stands for Least Absolute Shrinkage and Selection Operator. It’s a linear regression technique that helps with overfitting and feature selection at the same time.**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
X,y=make_regression(n_samples=100,n_features=1,n_informative=1,n_targets=1,noise=20,random_state=13)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
plt.scatter(X_train,y_train)
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
reg.fit(X_train,y_train)
print(reg.coef_)
print(reg.intercept_)
```

```python
alphas=[0,1,5,10,30]
plt.figure(figsize=(12,6))
plt.scatter(X,y)
for i in alphas:
  l=Lasso(alpha=i)
  l.fit(X_train,y_train)
  plt.plot(X_test,l.predict(X_test),label=f"alpha={i}")
plt.legend()
plt.show()
```

# **1.How are coeffcients affected?**

```python
from sklearn.datasets import load_diabetes
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
data=load_diabetes()
df=pd.DataFrame(data.data,columns=data.feature_names)
df["target"]=data.target
df.head()
```

```python
X_train,X_test,y_train,y_test=train_test_split(data.data,data.target,test_size=0.2,random_state=2)
```

```python
from sklearn.metrics import r2_score
coefs=[]
r2_scores_list=[]
for u in [0,0.1,1,10]:
  reg=Lasso(alpha=u)
  reg.fit(X_train,y_train)

  coefs.append(reg.coef_.tolist())
  y_pred=reg.predict(X_test)
  r2_scores_list.append(r2_score(y_test,y_pred))
```

```python
plt.figure(figsize=(14,9))
plt.subplot(221)
plt.bar(data.feature_names,coefs[0])
plt.title("Alpha=0, r2_score={}".format(round(r2_scores_list[0],2)))

plt.subplot(222)
plt.bar(data.feature_names,coefs[1])
plt.title("Alpha=0.1, r2_score={}".format(round(r2_scores_list[1],2)))

plt.subplot(223)
plt.bar(data.feature_names,coefs[2])
plt.title("Alpha=1, r2_score={}".format(round(r2_scores_list[2],2)))

plt.subplot(224)
plt.bar(data.feature_names,coefs[3])
plt.title("Alpha=10, r2_score={}".format(round(r2_scores_list[3],2)))

plt.show()
```

# **3.Impact on Bias and Variance**

```python
m=100
X2=5*np.random.rand(m,1)-2
y2=0.7*X**2 - 2 *X+3+np.random.randn(m,1)
plt.scatter(X2,y2)
plt.show()
```

```python
X_train,X_test,y_train,y_test=train_test_split(X2.reshape(100,1),y2.reshape(100),test_size=0.2,random_state=2)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=10)
x_train=poly.fit_transform(X_train)
x_test=poly.fit_transform(X_test)
```

```python
from mlxtend.evaluate import bias_variance_decomp
alphas=np.linspace(0,30,100)
loss=[]
bias=[]
variance=[]
for i in alphas:
  reg=Lasso(alpha=i)
  avg_expected_loss,avg_bias,avg_variance=bias_variance_decomp(reg, X_train,y_train,X_test,y_test,loss='mse',random_seed=123)
  loss.append(avg_expected_loss)
  bias.append(avg_bias)
  variance.append(avg_variance)
```

```python
plt.plot(alphas,loss,label="loss")
plt.plot(alphas,bias,label="bias")
plt.plot(alphas,variance,label="variance")
plt.xlabel("Alpha")
plt.ylabel("Error")
plt.legend()
plt.show()
```
