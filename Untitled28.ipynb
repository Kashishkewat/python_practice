# **FUNCTIONS IN PYTHON**

---



**1. What is a Function?**

A function is a block of reusable code that performs a specific task.

1.  Reduces code repetition
2.  Improves readability
3. Makes programs modular

**2.creating function**

```python
def function_name():
    # code
```

```python
def greet():
    print("Hello, Python!")

greet()   # function call
```

**3.Function with Parameters**

```python
def greet(name):
    print("Hello", name)

greet("Kashish")
```

**4.Function with Return Value**

```python
def add(a, b):
    return a + b

result = add(10, 5)
print(result)
```

**5.Function with Multiple Parameters**

```python
def student(name, age, course):
    print(name, age, course)

student("Aman", 21, "CS")
```

**6. Default Arguments**

```python
def greet(name="Student"):
    print("Hello", name)

greet()
greet("Riya")
```

**7.Keyword Arguments**

```python
def info(name, age):
    print(name, age)

info(age=22, name="Kashish")
```

**8. Variable-Length Arguments**

```python
def total(*nums):#*args
    return sum(nums)

print(total(1, 2, 3))
```

```python
def details(**info):#**kwargs
    print(info)

details(name="Aman", age=20)
```

**9. Lambda Function (Anonymous)**

```python
square = lambda x: x * x
print(square(5))
```

**10. Recursive Function**

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)

print(factorial(5))
```

# **Mini Project**

---



```python
print("checking prime number")
def is_prime(num):
    if num <= 1:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False
    return True


n = int(input("Enter a number: "))

if is_prime(n):
    print("Prime number")
else:
    print("Not a prime number")
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact


num = int(input("Enter a number: "))
print("Factorial =", factorial(num))
```
