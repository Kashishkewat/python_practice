# **OPERATORS IN PYTHON:**
Operators are used to perform operations on values and variables.

**1. Arithmetic Operators**

```python
a=3
b=9
print(a+b)#addition
print(a-b)#subraction
print(a*b)#Multiplication
print(a/b)#Division
print(a%b)#Modulus
print(a**b)#Power
print(a//b)#Floor Division
```

**2.Assignment Operators**

**3**. **Comparison** (**Relational**) **Operators**

```python
a = 10
b = 5
print(a > b)#Greater
print(a == b)#equal
print(a < b)#less
print(a != b)#not equal
print(a >= b)#greater than equal to
print(a <= b)#less than equal to
```

**4.Logical Operators**

```python
x = 10
print(x > 5 and x < 20)
print(x > 7 or x < 23)
print(not(x > 5))
```

**6**. **Membership** **Operators**

```python
nums = [1, 2, 3]
print(2 in nums) # if 2 is present in num it will print true or false
print(8 not in nums)
```

**7. Identity Operators**

```python
a = 10
b = 10
print(a is b)# if the  value of a is same as b it will print true
print(a is not b)# if the value of a is not same as b it will print true
```
