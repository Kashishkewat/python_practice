# **ERROR HANDLING IN PYTHON**

---



**1. What is an Error?**

An error (exception) occurs when the program stops due to a problem.

Examples:

1. ZeroDivisionError

2. ValueError

3. TypeError

4. FileNotFoundError

```python
#try_except block

try:
    a = int(input("Enter number: "))
    print(10 / a)
except:
    print("Error occurred")
```

```python
# Handling Specific Errors
try:
    x = int("abc")
except ValueError:
    print("Value error")
```

```python
#runs when no error occurs
try:
    num = int(input("Enter number: "))
except ValueError:
    print("Invalid input")
else:
    print("You entered:", num)
```

```python
#finally block(always runs either error or not )
try:
    print(10 / 2)
finally:
    print("Program finished")
```


# **FILE HANDLING IN PYTHON**


---



File handling is used to read and write data to files.

**1.opening a files**

```python
file = open("data.txt", "r")
```

**2. Writing to a File**

```python
file = open("data.txt", "w")
file.write("Hello Python")
file.close()
```

**3.Reading from a File**

```python
file = open("data.txt", "r")
content = file.read()
print(content)
file.close()
```

**4. Reading Line by Line**

```python
with open("data.txt", "r") as file:
    for line in file:
        print(line)
```

**5. Append to File**

```python
with open("data.txt", "a") as file:
    file.write("\nNew line added")
```

**6.File Handling with Error Handling**

```python
try:
    with open("data.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found")
```

```python
def calculate_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "Fail"


students = []   # list of dictionaries

while True:
    print("\n--- Student Report Card System ---")
    print("1. Add Student")
    print("2. Display Report Cards")
    print("3. Save to File")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    # 1. Add student
    if choice == 1:
        name = input("Enter student name: ")

        marks = []
        for i in range(1, 4):
            m = int(input(f"Enter marks of Subject {i}: "))
            marks.append(m)

        total = sum(marks)
        average = total / len(marks)
        grade = calculate_grade(average)

        student = {
            "name": name,
            "marks": marks,
            "total": total,
            "average": average,
            "grade": grade
        }

        students.append(student)
        print("Student added successfully!")

    # 2. Display report cards
    elif choice == 2:
        if not students:
            print("No student records found.")
        else:
            for s in students:
                print("\nName:", s["name"])
                print("Marks:", s["marks"])
    elif choice == 3:
        with open("report_card.txt", "w") as file:
            for s in students:
                file.write(f"Name: {s['name']}\n")
                file.write(f"Marks: {s['marks']}\n")
                file.write(f"Total: {s['total']}\n")
                file.write(f"Average: {s['average']}\n")
                file.write(f"Grade: {s['grade']}\n")
                file.write("-" * 30 + "\n")

        print("Report cards saved to file successfully!")


    elif choice == 4:
        print("Thank you! Program exited.")
        break

    else:
        print("Invalid choice. Try again.")
```
