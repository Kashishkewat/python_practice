# **Bagging Regressor**

**Core idea**

**Train many regression models on different random samples of the same dataset → average their predictions.**

**This reduces variance and makes predictions more stable (less overfitting).**

```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import BaggingRegressor
```

```python
cal = fetch_california_housing()
X_cal,Y_cal=cal.data,cal.target
print("dataset features names:" +str(cal.feature_names))
print("dataset features size:" +str(cal.data.shape))
print("dataset target names:" +str(cal.target.shape))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X_cal,Y_cal,train_size=0.80,test_size=0.20,random_state=123)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
print(Y_train.shape)
print(Y_test.shape)
```

```python
lr=LinearRegression()
dt=DecisionTreeRegressor()
knn=KNeighborsRegressor()
```

```python
lr.fit(X_train,Y_train)
dt.fit(X_train,Y_train)
knn.fit(X_train,Y_train)
```

```python
y_pred1=lr.predict(X_test)
y_pred2=dt.predict(X_test)
y_pred3=knn.predict(X_test)
#
```

```python
print("r2 score of lr",r2_score(Y_test,y_pred1))
print("r2 score of dt",r2_score(Y_test,y_pred2))
print("r2 score of knn",r2_score(Y_test,y_pred3))
```

```python
bag_reg=BaggingRegressor(random_state=42)
bag_reg.fit(X_train,Y_train)
```

```python
y_pred=bag_reg.predict(X_test)
print("traing coefficient of r2",bag_reg.score(X_train,Y_train))
print("test coefficient of r2",bag_reg.score(X_test,Y_test))
```

```python
n_sample=cal.data.shape[0]
n_features=cal.data.shape[1]

params={
    "estimator":[None,LinearRegression(),KNeighborsRegressor()],
     "n_estimators":[20,50,100],
    "max_samples":[0.5,1.0],
    "max_features":[0.5,1.0],
    "bootstrap":[True,False],
    "bootstrap_features":[True,False]
}
bag_reg_grid=GridSearchCV(BaggingRegressor(random_state=42),param_grid=params,cv=3,n_jobs=-1,verbose=1)
bag_reg_grid.fit(X_train,Y_train)
```

```python
print("train r2 score",bag_reg_grid.best_estimator_.score(X_train,Y_train))
print("test r2 score",bag_reg_grid.best_estimator_.score(X_test,Y_test))
print("best r2 score through grid search",bag_reg_grid.best_score_)
print("best parameters",bag_reg_grid.best_params_)
```
