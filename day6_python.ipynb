# **DICTIONARY IN PYTHON**

---



1.What is a Dictionary?

A dictionary stores data in key–value pairs.

1. Unordered (ordered from Python 3.7+)
2. Changeable (mutable)
3. Keys must be unique

```python
student = {
    "name": "Kashish",
    "age": 21,
    "course": "Computer Science"
}
```

**2.Creating a Dictionary**

```python
empty = {}

person = dict(name="Aman", age=22)
```

**3.Accessing Dictionary Values**

```python
print(student["name"])
print(student.get("age"))
```

**4.Adding and Updating Items**

```python
student["college"] = "ABC College"
student["age"] = 22
```

**5. Removing Items**

```python
student.pop("age")
student.popitem()     # removes last item
del student["course"]
```

**6. Looping Through Dictionary**

```python
for key in student:
    print(key, student[key])
```

```python
for k, v in student.items():
    print(k, v)
```

**7.Check Key Exists**

```python
if "name" in student:
    print("Key exists")
```

**8. Nested Dictionary**

```python
students = {
    "s1": {"name": "Aman", "age": 20},
    "s2": {"name": "Riya", "age": 21}
}

print(students["s1"]["name"])
```

**9.Dictionary Comprehension**

```python
squares = {x: x*x for x in range(1, 6)}
print(squares)
```

**10.Dictiontary Functions**

```python
student = {"name": "Kashish", "age": 21}#Returns value of a key
print(student.get("name"))
print(student.get("marks"))   # None
```

```python
print(student.values())#return all values
```

```python
print(student.items())# return key-value pair
```

```python
new_student = student.copy()# create the copy of existing dict
```

```python
student.setdefault("marks", 0)#set default for any key
```

```python
keys = ["a", "b", "c"]#Creates dictionary from keys with same value
new_dict = dict.fromkeys(keys, 0)
print(new_dict)
```

# **Mini Project**

---



```python
students = {}

while True:
    print("\n--- Student Marks Management System ---")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Calculate Result")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    # 1. Add student
    if choice == 1:
        name = input("Enter student name: ")
        marks1 = int(input("Enter marks in Subject 1: "))
        marks2 = int(input("Enter marks in Subject 2: "))
        marks3 = int(input("Enter marks in Subject 3: "))

        students[name] = [marks1, marks2, marks3]
        print("Student added successfully!")

    # 2. Display students
    elif choice == 2:
        if not students:
            print("No student data available.")
        else:
            for name, marks in students.items():
                print(f"{name} : {marks}")

    # 3. Calculate result
    elif choice == 3:
        name = input("Enter student name: ")

        if name in students:
            marks = students[name]
            total = sum(marks)
            average = total / 3

            if average >= 90:
                grade = "A"
            elif average >= 75:
                grade = "B"
            elif average >= 50:
                grade = "C"
            else:
                grade = "Fail"

            print("Total Marks:", total)
            print("Average:", average)
            print("Grade:", grade)
        else:
            print("Student not found.")

    # 4. Exit
    elif choice == 4:
        print("Thank you! Program exited.")
        break

    else:
        print("Invalid choice! Try again.")
```
