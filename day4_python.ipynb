What is a List?

A list is a collection of items that is:
1.  Ordered
2.  Changeable (mutable)
3.  Allows duplicate values

```python
numbers = [10, 20, 30, 40]
```

**2.Creating a List**

```python
empty_list = []
names = ["Kashish", "Aman", "Riya"]
mixed = [1, "Python", 3.5, True]
```

**3. Accessing List Elements**

```python
fruits = ["apple", "banana", "mango"]
print(fruits[0])     # apple
print(fruits[-1])    # mango
```

**4. Changing List Elements**

```python
fruits[1] = "orange"
print(fruits)
```

**5. List Slicing**

```python
numbers = [1, 2, 3, 4, 5]

print(numbers[1:4])   # [2, 3, 4]
print(numbers[:3])    # [1, 2, 3]
print(numbers[::2])   # [1, 3, 5]
```

**6. Looping Through a List**

```python
for item in nums:
    print(item)
```

```python
for i in range(len(nums)):#using index
    print(nums[i])
```

**7. List Comprehension (Short Way)**

```python
squares = [x*x for x in range(1, 6)]
print(squares)
```

**8.common list operation**

```python
a = [1, 2, 3]
b = [4, 5]

print(a + b)      # Join lists
print(a * 2)      # Repeat list
```

**9.Nested List**

```python
matrix = [
    [1, 2],
    [3, 4]
]

print(matrix[1][0])   # 3
```

**LIST FUNCTIONS (METHODS) IN PYTHON**
---


```python
nums = [1, 2, 3]#Adds one element at the end of the list
nums.append(4)
print(nums)
```

```python
nums.insert(1, 10)#Adds an element at a specific position
print(nums)
```

```python
nums.extend([5, 6])#Adds multiple elements from another list
print(nums)
```

```python
nums.remove(10)#Removes the first occurrence of an element
print(nums)
```

```python
nums.pop()#Removes and returns element by index
#(Default: last element)
nums.pop(1)
print(nums)
```

```python
nums.clear()#Removes all elements from the list
print(nums)
```

```python
nums = [10, 20, 30]#Returns the index of an element
print(nums.index(20))
```

```python
nums = [1, 2, 2, 3]#Counts how many times an element appears
print(nums.count(2))
```

```python
nums = [3, 1, 4, 2]#Sorts the list in ascending order
nums.sort()
print(nums)
```

```python
nums.sort(reverse=True)#DESCENDING ORDER
```

```python
nums.reverse()#Reverses the list order
print(nums)
```

```python
new_list = nums.copy()#Creates a copy of the list
```

```python
nums = [1, 2, 3]#returns number of elements
print(len(nums))
```

```python
nums = [5, 2, 9]#Find largest and smallest elements
print(max(nums))
print(min(nums))
```

```python
nums = [1, 2, 3]#Adds all numeric elements
print(sum(nums))
```
