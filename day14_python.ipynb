```python
import numpy as np

zeros_array = np.zeros((2, 3))#np.zeros(): Creates an array filled with zeros.
ones_array = np.ones((3, 3))#np.ones(): Creates an array filled with ones.
constant_array = np.full((2, 2), 7)#np.full():create an array filled with speificed value
range_array = np.arange(0, 10, 2)  # start, stop, step np.arange(): Creates an array with values that are evenly spaced within a given range.

linspace_array = np.linspace(0, 1, 5)  # start, stop, num

print("Zero Array:","\n",zeros_array)
print("Ones Array:","\n",ones_array)
print("Constant Array:","\n",constant_array)
print("Range Array:","\n",range_array)
print("Linspace Array:","\n",linspace_array)
```

**6.numpy.stack() in python**

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
res = np.stack((a, b), axis=0)#axis=0: a and b become rows.

print(res)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
res = np.stack((a, b), axis=1)#axis=1: a and b become columns.

print(res)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
res = np.stack((a, b), axis=-1)#axis=-1: same as axis 1 because it refers to the last dimension.

print(res)
```

```python
x = np.array([[1, 2, 3],
              [4, 5, 6]])

y = np.array([[7, 8, 9],
              [10, 11, 12]])

print(np.stack((x, y), axis=0))
print(np.stack((x, y), axis=1))
print(np.stack((x, y), axis=2))
```

```python

m = np.array([[[1, 2], [3, 4]],
              [[5, 6], [7, 8]]])

n = np.array([[[10, 20], [30, 40]],
              [[50, 60], [70, 80]]])

print(np.stack((m, n), axis=0))
print(np.stack((m, n), axis=1))
print(np.stack((m, n), axis=2))
print(np.stack((m, n), axis=3))
```

**Matrix Operations in NumPy**


```python

x = np.array([[1, 2], [4, 5]])
y = np.array([[7, 8], [9, 10]])

print("Addition:\n", np.add(x, y))
print("Subtraction:\n", np.subtract(x, y))
print("Division:\n", np.divide(x, y))
```

```python
print("Element-wise multiplication:\n", np.multiply(x, y))
```

```python
A = np.array([[1, 2],
              [3, 4]])

res = np.dot(A, A)    # Matrix multiplication
print(res)
```

```python
x = np.array([[1, 2], [4, 5]])
y = np.array([[7, 8], [9, 10]])

print("Square root:\n", np.sqrt(x))
print("Sum of all elements:", np.sum(y))

print("Column-wise sum:", np.sum(y, axis=0))
print("Row-wise sum:", np.sum(y, axis=1))
print("Transpose:\n", x.T)
```

**Products of Vectors and Matrices in NumPy**


```python
import numpy as np

# Define vectors
a = np.array([2, 6])#For vectors a and b: (2 * 3) + (6 * 10) = 66

b = np.array([3, 10])

print("Inner product of vectors a and b =")
print(np.inner(a, b))

# Define matrices
x = np.array([[2, 3, 4], [3, 2, 9]])
y = np.array([[1, 5, 0], [5, 10, 3]])

print("Inner product of matrices x and y =")
print(np.inner(x, y))
```

```python
# Define vectors
a = np.array([2, 6])
b = np.array([3, 10])

print("Outer product of vectors a and b =")#Example: First row: [2 * 3, 2 * 10] = [6, 10]
print(np.outer(a, b))

# Define matrices
x = np.array([[3, 6, 4], [9, 4, 6]])
y = np.array([[1, 15, 7], [3, 10, 8]])

print("Outer product of matrices x and y =")
print(np.outer(x, y))
```

```python
# Define vectors as 3D (z=0 for 2D compatibility)
a = np.array([3, 6, 0])
b = np.array([9, 10, 0])

print("Cross product of vectors a and b =")
print(np.cross(a, b))

# Define matrices (already 3D rows)
x = np.array([[2, 6, 9], [2, 7, 3]])
y = np.array([[7, 5, 6], [3, 12, 3]])

print("Cross product of matrices x and y =")
print(np.cross(x, y))
```
