```python
# simple calculator
print("Simple Calculator")
num1=int(input("enter first number:"))
num2=int(input("enter second number:"))
operator=input("enter operator:")

if operator=="+":
    print(num1+num2)
elif operator=="-":
  print(num1-num2)
elif operator=='*':
  print(num1*num2)
elif operator=='/':
  print(num1/num2)
else:
  print("invalid operator")
```
