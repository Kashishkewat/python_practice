# **TUPLES IN PYTHON**

---



**1. What is a Tuple?**

A tuple is a collection of items that is:
1. Ordered
2. Unchangeable (immutable)
3. Allows duplicate values

```python
t = (10, 20, 30)
```

**2. Creating a Tuple**

```python
empty = ()
single = (5,)      # comma is required
mixed = (1, "Python", 3.5)
```

**3. Accessing Tuple Elements**

```python
colors = ("red", "green", "blue")
print(colors[0])
print(colors[-1])
```

**4.Tuple Slicing**

```python
nums = (1, 2, 3, 4, 5)
print(nums[1:4])
```

**5.Tuple Function**

```python
nums = (1, 2, 2, 3)
print(nums.count(2))
print(nums.index(3))
```

**6.Tuple Operation**

```python
a = (1, 2)
b = (3, 4)

print(a + b)     # Concatenation
print(a * 2)     # Repetition
```

# **SETS IN PYTHON**

---



**1.What is a Set?**
**bold text**
A set is a collection of items that is:
1. Unordered
2. Changeable
3. No duplicate values

```python
s = {1, 2, 3}
```

**2. Creating a Set**

```python
empty = set()
nums = {1, 2, 2, 3}
print(nums)    # duplicates removed
```

**3. Adding and Removing Elements**

```python
nums = {1, 2, 2, 3}

nums.add(4)
# nums.remove(2)
# nums.discard(10)   # no error if not found
print(nums)
```

```python
nums.remove(2)
print(nums)
```

```python
nums.discard(10)   # no error if not found
print(nums)
```

**4. Set Operations**

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a | b)   # Union
print(a & b)   # Intersection
print(a - b)   # Difference
print(a ^ b)   # Symmetric Difference
```

```python
nums = [1, 2, 3, 2, 4, 1, 5]
unique_nums = []

for i in nums:
    if i not in unique_nums:
        unique_nums.append(i)

print(unique_nums)
```
