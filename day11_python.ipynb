# **Polymorphism**

---



**1. Polymorphism with Functions**

```python
#The same function works with different data types.
print(len("Python"))     # string
print(len([1, 2, 3, 4])) # list
print(len((10, 20)))     # tuple
```

**2.Polymorphism with Operators**

```python
#The same operator performs different operations.
print(5 + 3)          # Addition
print("Py" + "thon")  # String concatenation
print([1, 2] + [3])   # List merging
```

**3. Polymorphism using Method Overriding (Inheritance)**

```python
#Child class redefines the method of the parent class.
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

a = Animal()
d = Dog()
c = Cat()

a.speak()
d.speak()
c.speak()
```

**4. Polymorphism using Duck Typing**

```python
#Python doesn’t care about class type, only about method availability.
class Laptop:
    def code(self):
        print("Coding on laptop")

class Desktop:
    def code(self):
        print("Coding on desktop")

def start_work(device):
    device.code()

l = Laptop()
d = Desktop()

start_work(l)
start_work(d)
```

**5.Polymorphism using Method Overloading (Not direct in Python)**

```python
#Python does not support traditional overloading, but we can simulate it.
class Math:
    def add(self, a, b=0, c=0):
        return a + b + c

m = Math()
print(m.add(2))
print(m.add(2, 3))
print(m.add(2, 3, 4))
```

**Mini Project**

```python
#1. Base Class: Employee
class Employee:
    def get_role(self):
        print("I am an Employee")

    def get_salary(self):
        print("Base salary of employee")

#2.Child Class: Manager (inherits Employee)
class Manager(Employee):
    def get_role(self):
        print("I am a Manager")

    def get_salary(self):
        print("Manager salary: Base + Bonus")

#3.Child Class: Developer (inherits Manager)
class Developer(Manager):
    def get_role(self):
        print("I am a Developer")

    def get_salary(self):
        print("Developer salary: Base + Project Allowance")

#4.polymorphism in Action

#Same method name → different behavior

employees = [Employee(), Manager(), Developer()]

for emp in employees:
    emp.get_role()
    emp.get_salary()
    print("------")
```
