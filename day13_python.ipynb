# **Inbuilt Module Python**

Inbuilt modules are pre-installed modules that come with Python.
You do not need to install them separately.

**1.math Module**

```python
import math   #Used for mathematical operations.

print(math.sqrt(16))
print(math.pow(2, 3))
print(math.factorial(5))
print(math.pi)
```

**2.random Module**

```python
import random   #Used to generate random values.

print(random.randint(1, 10))
print(random.choice([10, 20, 30]))
print(random.random())
```

**3.datetime Module**

```python
from datetime import datetime #Used to work with date and time.

now = datetime.now()
print(now)
print(now.date())
print(now.time())
```

**4.os Module**

```python
#Used to interact with the operating system.
import os

print(os.getcwd())      # current directory
os.mkdir("test_folder") # create folder
```

**5.sys Module**

```python
import sys
#Used for system-specific operations.
print(sys.version)
print(sys.argv)
```

**6.time Module**

```python
import time #Used for time-related operations.

print(time.time())
time.sleep(2)
print("After 2 seconds")
```

**7.calendar Module**

```python
#Used to display calendar information.
import calendar

print(calendar.month(2026, 1))
```

**8.statistics Module**

```python
#Used for statistical calculations.
import statistics

data = [10, 20, 30, 40]
print(statistics.mean(data))
print(statistics.median(data))
```

# **Mini Project**

---



```python
import os

os.makedirs('math_utils', exist_ok=True)
with open("math_utils/__init__.py", "w") as f:
    f.write("# Math Utilities Package\n")
    f.write("from . import basic\n")
    f.write("from . import advanced\n")
    f.write("from . import stats\n")
```

```python
with open("math_utils/basic.py", "w") as f:
    f.write("""
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b
""")
```

```python
with open("math_utils/advanced.py", "w") as f:
    f.write("""
def power(base, exp):
    return base ** exp

def factorial(n):
    if n < 0:
        return "Invalid number"
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
""")
```

```python
with open("math_utils/stats.py", "w") as f:
    f.write("""
def mean(numbers):
    return sum(numbers) / len(numbers)

def maximum(numbers):
    return max(numbers)

def minimum(numbers):
    return min(numbers)
""")
```

```python
from math_utils import basic, advanced, stats

print(basic.add(10, 5))
print(advanced.factorial(5))
print(stats.mean([10, 20, 30]))
```
