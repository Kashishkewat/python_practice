# **Overfitting-Underfitting in Decision Tree**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
def analyzer(max_depth):
  df = pd.read_csv('Social_Network_Ads.csv')
  # Select 'Age' (index 2) and 'EstimatedSalary' (index 3) as features
  X = df.iloc[:, [2, 3]].values
  y = df.iloc[:, -1].values
  clf=DecisionTreeClassifier(max_depth=max_depth)
  clf.fit(X,y)
  a=np.arange(start=X[:,0].min()-1,stop=X[:,0].max()+1,step=1) # Increased step size
  b=np.arange(start=X[:,1].min()-1,stop=X[:,1].max()+1,step=1) # Increased step size
  XX,YY=np.meshgrid(a,b)
  input_array=np.array([XX.ravel(),YY.ravel()]).T
  Z=clf.predict(input_array)
  plt.contourf(XX,YY,Z.reshape(XX.shape),alpha=0.5)
  plt.scatter(X[:,0],X[:,1],c=y,cmap='rainbow')
  plt.show()
```

```python
analyzer(max_depth=3)
#
```
