## **Gradient descent**

### **Gradient Descent is an optimization algorithm used in Machine Learning to minimize the error (loss) of a model by updating its parameters step by step.**

```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y=make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

```python
#Lets apply OLS
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(X,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
#lets apply gradient descent assuming slope is constant m=78.35
# and let's assume the starting value for intercept b=0
y_pred=((78.35*X)+0).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label="OLS")
plt.plot(X,y_pred,color='green',label="Gradient Descent")
plt.legend()
plt.show()
```

```python
m=78.35
b=0
loss_slope=-2 * np.sum(y-m*X.ravel()-b)
loss_slope
```

```python
# lets take learning rate=0.1
learning_rate=0.1
step_size=loss_slope*learning_rate
step_size
```

```python
b=b - step_size
b
```

```python
y_pred1=((78.35 * X )+b).reshape(4)
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label="OLS")
plt.plot(X,y_pred,color='green',label="b = 0")
plt.plot(X,y_pred1,color='blue',label="b = {}".format(b))

plt.legend()
plt.show()
```

```python
#Iteration 2
loss_slope=-2 * np.sum(y-m*X.ravel()-b)
loss_slope
```

```python
# lets take learning rate=0.1
learning_rate=0.1
step_size=loss_slope*learning_rate
step_size
```

```python
b=b - step_size
b
```

```python
y_pred2=((78.35 * X )+b).reshape(4)
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red',label="OLS")
plt.plot(X,y_pred,color='green',label="b = 0")
plt.plot(X,y_pred1,color='blue',label="b = {}".format(b))
plt.plot(X,y_pred2,color='black',label="b = {}".format(b))

plt.legend()
plt.show()
```

```python
#Iteration 3
loss_slope=-2 * np.sum(y-m*X.ravel()-b)
loss_slope
```

```python
# lets take learning rate=0.1
learning_rate=0.1
step_size=loss_slope*learning_rate
step_size
```

```python
b=b - step_size
b
```

```python
# gradient descent in linear regression
class GDRegressor:
  def __init__(self,learning_rate=0.01,epochs=100):
    self.learning_rate=learning_rate
    self.m=29.19
    self.b=-120
    self.epochs=epochs
  def fit(self,X_train,y_train):
    #calculate the b using GD
    for i in range(self.epochs):
      loss_slope=-2*np.sum(y_train-self.m*X_train.ravel()-self.b)
      self.b=self.b-(self.learning_rate*loss_slope)
    print(self.b)
```

```python
gd=GDRegressor(0.01,10)
```

```python
gd.fit(X2,y2)
```
