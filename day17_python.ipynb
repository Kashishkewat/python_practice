```python
# violin plot
sns.violinplot(data=penguins,x="species",y="body_mass_g",hue="sex",palette="Set1")
sns.swarmplot(data=penguins,x="species",y="body_mass_g",color="black")
```

KDE-> for estimmating the probability density function of continuous or non parametric data
or in simple terms: a smooth curve that shows the distrabution of data - like a smoothed -out version .It hepls you see where the data is concentrated(peaks) and how it spreads.

```python
sns.kdeplot(data=penguins,x="body_mass_g",hue="species",palette="pastel",fill="True")
sns.despine()
```

**heatmap: a 2d color_coded matrix that shows relationshils or patterns between two variables.**

**It's commonly used to:
-:show correlation between features**

**-:visualize confusion matrics-> will see in the later stages.**

**-:highlight missing data->sns.heatmap(df.isnull())**

```python
columns=['bill_length_mm',"bill_depth_mm","flipper_length_mm","body_mass_g"]
sns.heatmap(data=penguins[columns].corr(),annot=True,cmap="Blues",vmin=-0.2,linewidth=2)
```

```python
#rug plot:
sns.rugplot(data=penguins,x="body_mass_g",y="species",palette="Set1",height=0.3)
```

```python
# pair plot:
sns.pairplot(data=penguins[columns],palette="Set1",diag_kind="kde")
```

```python
#pair grid-> subplot grid for plotting pairwise relationships in a dataset.
plot_columns = columns + ['sex']
g=sns.PairGrid(data=penguins[plot_columns],hue="sex",palette="Set1")
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
g.map_diag(sns.histplot)
```

# **Mini Project Python**

```python
#Bar Chart
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
subjects = ['Math', 'Science', 'English', 'Computer']
marks = [85, 90, 78, 92]

sns.barplot(x=subjects, y=marks)
plt.title("Bar Chart - Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
# Line Graph
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
days = [1, 2, 3, 4, 5]
sales = [200, 250, 300, 280, 350]

sns.lineplot(x=days, y=sales)
plt.title("Line Graph - Sales Over Days")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.show()
```

```python
#Pie Chart
import seaborn as sns
import matplotlib.pyplot as plt

# Seaborn style
sns.set_style("whitegrid")

labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 25, 20, 15]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart - Programming Language Usage")
plt.axis('equal')  # Makes circle perfect
plt.show()
```
