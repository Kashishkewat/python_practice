# **CONTROL FLOW IN PYTHON**
Control flow statements include:
1. Conditional statements
2. Looping statements
3. Loop control statements





# **1.Conditional statements**

---


# if statement
Executes code only if condition is True**

```python
age = 18

if age >= 18:
    print("You are eligible to vote")
```

# **if–else statement**

```python
num = 5

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

# **if–elif–else statement**


```python
#Used for multiple conditions
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

# **Nested if**

```python
num = 10

if num > 0:
    if num % 2 == 0:
        print("Positive Even")
```

# **2.Looping Statements**

---


Loops are used to repeat code

**1. for loop**
=>Used when number of iterations is known

```python
for i in range(1, 6):
    print(i)
```

**2.while loop**=>Runs while condition is True

```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

**3**. **Loop** **with** **else**

```python
for i in range(3):
    print(i)
else:
    print("Loop completed")
```

# **3.Loop Control Statements**

---



# **1.break(Stops loop immediately)**

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

# **2.continue(Skips current iteration)**

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

# **3. pass(Does nothing (placeholder))**

```python
if True:
    pass
```

# **4**.**Match** **Case**

---





```python
day = "c"

match day:
    case "a":
        print("Monday")
    case "b":
        print("Tuesday")
    case "c":
        print("Wednesday")
    case _:
        print("Invalid day")
```
