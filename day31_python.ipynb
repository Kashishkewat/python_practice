# **Accuracy and Confusion matrix**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import files
uploaded=files.upload()
```

```python
df=pd.read_csv("heart.csv")
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,0:-1],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf1=DecisionTreeClassifier()
clf2=LogisticRegression()
```

```python
clf1.fit(X_train,y_train)
clf2.fit(X_train,y_train)
```

```python
y_pred1=clf1.predict(X_test)
y_pred2=clf2.predict(X_test)
#
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
print("accracy of logistic regression",accuracy_score(y_test,y_pred2))
print("accracy of descision trees",accuracy_score(y_test,y_pred1))
```

```python
print("confusion matrix of logistic regression",confusion_matrix(y_test,y_pred2))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred1))
```

```python
from sklearn.datasets import load_iris
```

```python
df1=load_iris()
```

```python
df_iris = pd.DataFrame(df1.data, columns=df1.feature_names)
df_iris['target'] = df1.target
df_iris.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
X_train,X_test,y_train,y_test=train_test_split(df_iris.iloc[:,0:-1],df_iris.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf_iris1=DecisionTreeClassifier()
clf_iris2=LogisticRegression()
```

```python
clf_iris1.fit(X_train,y_train)
clf_iris2.fit(X_train,y_train)
```

```python
y_pred3=clf_iris1.predict(X_test)
y_pred4=clf_iris2.predict(X_test)
#
```

```python
print("accracy of logistic regression",accuracy_score(y_test,y_pred4))
print("accracy of descision trees",accuracy_score(y_test,y_pred3))
```

```python
print("confusion matrix of logistic regression",confusion_matrix(y_test,y_pred4))
```

```python
print(classification_report(y_test,y_pred4))
```
