# **Accuracy and Confusion matrix**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import files
uploaded=files.upload()
```

```python
df=pd.read_csv("heart.csv")
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,0:-1],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf1=DecisionTreeClassifier()
clf2=LogisticRegression()
```

```python
clf1.fit(X_train,y_train)
clf2.fit(X_train,y_train)
```

```python
y_pred1=clf1.predict(X_test)
y_pred2=clf2.predict(X_test)
#
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
print("accracy of logistic regression",accuracy_score(y_test,y_pred2))
print("accracy of descision trees",accuracy_score(y_test,y_pred1))
```

```python
print("confusion matrix of logistic regression",confusion_matrix(y_test,y_pred2))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred1))
```

```python
X_train,X_test,y_train,y_test=train_test_split(df2.iloc[:,1:],df2.iloc[:,0],test_size=0.2,random_state=2)
```

```python
from sklearn.metrics import precision_score,recall_score,f1_score
```

```python
model=LogisticRegression()
model.fit(X_train,y_train)
```

```python
y_pred5=model.predict(X_test)
#
```

```python
print("accracy of logistic regression",accuracy_score(y_test,y_pred5))
```

```python
print("confusion matrix of logistic regression",confusion_matrix(y_test,y_pred5))
```

```python
print(precision_score(y_test,y_pred5,average=None))
print(recall_score(y_test,y_pred5,average=None))
```

```python
print(f1_score(y_test,y_pred5,average=None))
```

```python
print(precision_score(y_test,y_pred5,average="macro"))#average of all precision
print(recall_score(y_test,y_pred5,average="macro"))
print(f1_score(y_test,y_pred5,average="macro"))
```

```python
print(precision_score(y_test,y_pred5,average="weighted"))
print(recall_score(y_test,y_pred5,average="weighted"))
print(f1_score(y_test,y_pred5,average="weighted"))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred5))
```
